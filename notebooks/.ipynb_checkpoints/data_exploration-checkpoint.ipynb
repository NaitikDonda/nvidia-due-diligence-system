```python
print(revenue_df["form"].value_counts())
print(net_income_df["form"].value_counts())
```

```python
revenue_annual = revenue_df[revenue_df["form"] == "10-K"].copy()
net_income_annual = net_income_df[net_income_df["form"] == "10-K"].copy()
```

```python
print(revenue_annual[["fy", "fp", "form", "val"]].tail(10))
print(net_income_annual[["fy", "fp", "form", "val"]].tail(10))
```

```python
revenue_annual = revenue_annual[["fy", "fp", "filed", "val"]]
net_income_annual = net_income_annual[["fy", "fp", "filed", "val"]]
```

```python
revenue_annual = revenue_annual.rename(columns={"val": "revenue"})
net_income_annual = net_income_annual.rename(columns={"val": "net_income"})
```

```python
financial_df = pd.merge(
    revenue_annual,
    net_income_annual,
    on=["fy", "fp", "filed"],
    how="inner"
)
```

```python
financial_df.tail(10)
```

```python
financial_df["profit_margin"] = (
    financial_df["net_income"] / financial_df["revenue"]
) * 100
```

```python
financial_df = financial_df.sort_values("fy").reset_index(drop=True)

financial_df["revenue_growth"] = (
    financial_df["revenue"].pct_change() * 100
)
```

```python
news_df.shape
```

```python
news_df.isnull().sum().sort_values(ascending=False)
```

```python
news_df.duplicated().sum()
```

```python
news_df = (
    news_df
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
news_df.to_csv(
    "../data/processed/nvidia_news_cleaned.csv",
    index=False
)
```

```python
news_df.shape
```

```python
date_columns = [
    col for col in news_df.columns
    if "date" in col.lower()
]

for col in date_columns:
    news_df[col] = pd.to_datetime(
        news_df[col],
        errors="coerce"
    )
```

```python
if date_columns:
    news_df = news_df.sort_values(
        by=date_columns[0]
    ).reset_index(drop=True)
```

```python
text_columns = news_df.select_dtypes(
    include="object"
).columns.tolist()

text_columns
```

```python
for col in text_columns:
    news_df[col] = (
        news_df[col]
        .astype(str)
        .str.strip()
    )
```

```python
news_df = news_df.replace(
    r"^\s*$",
    pd.NA,
    regex=True
)
```

```python
news_df.isnull().sum().sort_values(ascending=False)
```

```python
news_df.to_csv(
    "../data/processed/nvidia_news_cleaned.csv",
    index=False
)
```

```python
news_df.shape
```

```python
possible_text_columns = [
    col for col in news_df.columns
    if any(word in col.lower() for word in ["headline", "title", "news", "article", "text"])
]

possible_text_columns
```

```python
news_text_column = possible_text_columns[0]
news_text_column
```

```python
news_df["clean_text"] = (
    news_df[news_text_column]
    .fillna("")
    .astype(str)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
```

```python
news_df = news_df[
    news_df["clean_text"].str.len() > 0
].reset_index(drop=True)
```

```python
news_df["text_length"] = news_df["clean_text"].str.len()
```

```python
news_df["word_count"] = (
    news_df["clean_text"]
    .str.split()
    .str.len()
)
```

```python
news_df = news_df[
    news_df["word_count"] >= 3
].reset_index(drop=True)
```

```python
news_df.to_csv(
    "../data/processed/nvidia_news_features.csv",
    index=False
)
```

```python
news_df.shape
```

```python
news_df[
    ["clean_text", "text_length", "word_count"]
].head(10)
```

```python
if date_columns:
    news_df[date_columns[0]].agg(["min", "max"])
```

```python
if date_columns:
    news_df["Year"] = news_df[date_columns[0]].dt.year

    news_df["Year"].value_counts().sort_index()
```

```python
sentiment_columns = [
    col for col in news_df.columns
    if "sentiment" in col.lower()
    or "polarity" in col.lower()
]

sentiment_columns
```

```python
for col in sentiment_columns:
    print(f"\n--- {col} ---")
    print(news_df[col].value_counts(dropna=False).head(20))
```

```python
stock_model_df = nvidia_stock.copy()

stock_model_df["Date"] = pd.to_datetime(
    stock_model_df["Date"],
    utc=True,
    errors="coerce"
)

stock_model_df["Year"] = stock_model_df["Date"].dt.year
stock_model_df["Month"] = stock_model_df["Date"].dt.month
stock_model_df["Day"] = stock_model_df["Date"].dt.day
```

```python
stock_model_df["Future_Return_20D"] = (
    stock_model_df["Close"].shift(-20)
    / stock_model_df["Close"]
    - 1
) * 100
```

```python
stock_model_df["Target"] = (
    stock_model_df["Future_Return_20D"] > 0
).astype(int)
```

```python
stock_model_df = stock_model_df.dropna(
    subset=["Future_Return_20D"]
).reset_index(drop=True)
```

```python
stock_model_df["Price_Change"] = (
    stock_model_df["Close"]
    - stock_model_df["Open"]
)

stock_model_df["Price_Change_Pct"] = (
    stock_model_df["Close"]
    / stock_model_df["Open"]
    - 1
) * 100

stock_model_df["High_Low_Range"] = (
    stock_model_df["High"]
    - stock_model_df["Low"]
)

stock_model_df["Close_Open_Ratio"] = (
    stock_model_df["Close"]
    / stock_model_df["Open"]
)
```

```python
stock_model_df["Volume_Change"] = (
    stock_model_df["Volume"].pct_change() * 100
)

stock_model_df["Volume_MA_20"] = (
    stock_model_df["Volume"]
    .rolling(window=20)
    .mean()
)
```

```python
stock_model_df["Momentum_10"] = (
    stock_model_df["Close"]
    - stock_model_df["Close"].shift(10)
)

stock_model_df["Momentum_20"] = (
    stock_model_df["Close"]
    - stock_model_df["Close"].shift(20)
)
```

```python
def predict_nvidia_stock_direction(
    model,
    scaler,
    data
):
    features = data[model_features].copy()
    features_scaled = scaler.transform(features)

    prediction = model.predict(features_scaled)
    probability = model.predict_proba(features_scaled)

    result = data[["Date", "Close"]].copy()
    result["Prediction"] = prediction
    result["Probability_Down"] = probability[:, 0]
    result["Probability_Up"] = probability[:, 1]

    result["Signal"] = result["Prediction"].map({
        0: "DOWN",
        1: "UP"
    })

    result["Confidence"] = probability.max(axis=1)

    return result
```

```python
risk_start = risk_text.lower().find(
    "risk factors"
)

risk_start
```

```python
risk_section = risk_text[
    risk_start:
    risk_start + 100000
]

print(risk_section[:10000])
```

```python
risk_keywords = [
    "risk",
    "regulatory",
    "export",
    "geopolitical",
    "supply chain",
    "competition",
    "cybersecurity",
    "litigation",
    "lawsuit",
    "intellectual property",
    "customer",
    "manufacturing",
    "depend",
    "restriction"
]
```

```python
import re

sentences = re.split(
    r"(?<=[.!?])\s+",
    risk_section
)

risk_sentences = [
    sentence.strip()
    for sentence in sentences
    if any(
        keyword in sentence.lower()
        for keyword in risk_keywords
    )
]

len(risk_sentences)
```

```python
risk_df = pd.DataFrame({
    "text": risk_sentences
})
```

```python
risk_df = (
    risk_df
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
def assign_risk_category(text):
    text_lower = text.lower()

    if any(word in text_lower for word in [
        "export",
        "government",
        "regulatory",
        "restriction"
    ]):
        return "Regulatory"

    if any(word in text_lower for word in [
        "supply chain",
        "manufacturing",
        "supplier",
        "production"
    ]):
        return "Operational"

    if any(word in text_lower for word in [
        "competition",
        "competitor",
        "market"
    ]):
        return "Market"

    if any(word in text_lower for word in [
        "lawsuit",
        "litigation",
        "intellectual property"
    ]):
        return "Legal"

    if any(word in text_lower for word in [
        "cybersecurity",
        "cyber",
        "security"
    ]):
        return "Cybersecurity"

    if any(word in text_lower for word in [
        "geopolitical",
        "china",
        "international"
    ]):
        return "Geopolitical"

    return "Other"
```

```python
risk_df["Risk_Category"] = (
    risk_df["text"].apply(assign_risk_category)
)
```

```python
def predict_risk(
    texts,
    vectorizer,
    model
):
    vectors = vectorizer.transform(texts)

    predictions = model.predict(vectors)

    if hasattr(model, "predict_proba"):
        probabilities = model.predict_proba(vectors).max(axis=1)
    else:
        probabilities = [None] * len(predictions)

    results = pd.DataFrame({
        "Text": texts,
        "Risk_Prediction": predictions,
        "Confidence": probabilities
    })

    return results
```

```python
valuation_input_df = financial_model_df[
    [
        "fy",
        "revenue",
        "net_income",
        "Revenue_Growth",
        "Net_Income_Growth",
        "profit_margin"
    ]
].copy()

valuation_input_df.to_csv(
    valuation_data_path,
    index=False
)
```

```python
valuation_input_df
```

```python
valuation_input_df["Revenue_Growth"] = (
    valuation_input_df["Revenue_Growth"]
    .replace([float("inf"), -float("inf")], pd.NA)
)

valuation_input_df["Net_Income_Growth"] = (
    valuation_input_df["Net_Income_Growth"]
    .replace([float("inf"), -float("inf")], pd.NA)
)

valuation_input_df = valuation_input_df.dropna().reset_index(
    drop=True
)
```

```python
average_revenue_growth = (
    valuation_input_df["Revenue_Growth"]
    .mean()
)

average_net_income_growth = (
    valuation_input_df["Net_Income_Growth"]
    .mean()
)

print("Average Revenue Growth:", average_revenue_growth)
print("Average Net Income Growth:", average_net_income_growth)
```

```python
latest_financial = valuation_input_df.iloc[-1]

latest_revenue = latest_financial["revenue"]
latest_net_income = latest_financial["net_income"]
latest_profit_margin = latest_financial["profit_margin"]

print("Latest Revenue:", latest_revenue)
print("Latest Net Income:", latest_net_income)
print("Latest Profit Margin:", latest_profit_margin)
```

```python
forecast_years = list(range(1, 6))

forecast_df = pd.DataFrame({
    "Year": forecast_years
})
```

```python
forecast_df["Revenue"] = [
    latest_revenue *
    (1 + average_revenue_growth / 100) ** year
    for year in forecast_df["Year"]
]
```

```python
forecast_df["Net_Income"] = [
    latest_net_income *
    (1 + average_net_income_growth / 100) ** year
    for year in forecast_df["Year"]
]
```

```python
forecast_df
```

```python
forecast_df.to_csv(
    "../data/processed/nvidia_valuation_forecast.csv",
    index=False
)
```

```python
discount_rate = 0.10

forecast_df["Discount_Factor"] = (
    1 / (1 + discount_rate) ** forecast_df["Year"]
)
```

```python
forecast_df["Discounted_Net_Income"] = (
    forecast_df["Net_Income"]
    * forecast_df["Discount_Factor"]
)
```

```python
terminal_growth_rate = 0.03

terminal_net_income = forecast_df["Net_Income"].iloc[-1]

terminal_value = (
    terminal_net_income
    * (1 + terminal_growth_rate)
    / (discount_rate - terminal_growth_rate)
)

terminal_value
```

```python
terminal_discount_factor = forecast_df[
    "Discount_Factor"
].iloc[-1]

present_terminal_value = (
    terminal_value
    * terminal_discount_factor
)

present_terminal_value
```

```python
present_value_forecast = (
    forecast_df["Discounted_Net_Income"].sum()
)

enterprise_value = (
    present_value_forecast
    + present_terminal_value
)

enterprise_value
```

```python
valuation_summary = pd.DataFrame({
    "Metric": [
        "Present Value of Forecast",
        "Present Terminal Value",
        "Enterprise Value",
        "Discount Rate",
        "Terminal Growth Rate"
    ],
    "Value": [
        present_value_forecast,
        present_terminal_value,
        enterprise_value,
        discount_rate,
        terminal_growth_rate
    ]
})

valuation_summary
```

```python
valuation_metrics.to_csv(
    "../data/processed/final_valuation_metrics.csv",
    index=False
)
```

```python
with open(
    "../data/processed/nvidia_research_corpus.txt",
    "r",
    encoding="utf-8"
) as f:
    research_text = f.read()
```

```python
research_sentences = re.split(
    r"(?<=[.!?])\s+",
    research_text
)

research_sentences = [
    sentence.strip()
    for sentence in research_sentences
    if len(sentence.strip()) > 30
]

len(research_sentences)
```

```python
research_df = pd.DataFrame({
    "text": research_sentences
})
```

```python
research_df = (
    research_df
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
research_vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=3,
    max_df=0.90,
    sublinear_tf=True
)

research_tfidf = research_vectorizer.fit_transform(
    research_df["text"]
)
```

```python
research_tfidf.shape
```

```python
from sklearn.cluster import KMeans

research_topic_model = KMeans(
    n_clusters=6,
    random_state=42,
    n_init=20
)

research_df["Topic"] = research_topic_model.fit_predict(
    research_tfidf
)
```
